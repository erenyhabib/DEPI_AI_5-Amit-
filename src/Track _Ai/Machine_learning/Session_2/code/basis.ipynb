```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Salary_Data.csv')
```

```python
df.head(5)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.scatter(df['YearsExperience'],df['Salary'])
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Years of Experience vs Salary')
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
x_train.shape ,x_test.shape ,y_train.shape ,y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
linear_reg=LinearRegression()
linear_reg.fit(x_train,y_train)
linear_reg.predict(x_test)
```

```python
plt.scatter(x_train,y_train,color='red' ,label='Training set')
plt.plot(x_train,linear_reg.predict(x_train),color='blue',label='Regression line')
plt.title('Training set')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

```python
plt.scatter(x_test,y_test,color='red' ,label='Test set')
plt.plot(x_test,linear_reg.predict(x_test),color='blue',label='Regression line')
plt.title('Test set')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
y_pred=linear_reg.predict(x_test)
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
```

```python
mse
r2
mae
```

```python
score=linear_reg.score(x_test,y_test)
score
```
