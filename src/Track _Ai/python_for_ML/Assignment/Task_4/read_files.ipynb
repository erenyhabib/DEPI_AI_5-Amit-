```python
class TextFileReader:
    """
    Reads a text file and counts its lines, words, and characters.
    """

    def __init__(self, file_path):
        self.file_path = file_path
        self.content = ""

    def read_file(self):
        """
        Reads the file and stores its content.
        """
        with open(self.file_path, "r") as file:
            self.content = file.read()

    def count_lines(self):
        """
        Returns the number of lines in the file.
        """
        return len(self.content.splitlines())

    def count_words(self):
        """
        Returns the total number of words in the file.
        """
        return len(self.content.split())

    def count_characters(self):
        """
        Returns the total number of characters in the file.
        """
        return len(self.content)

    def display_content(self):
        """
        Displays the content of the file.
        """
        print(self.content)
```

```python
reader = TextFileReader("sample.txt")

reader.read_file()

print("Lines:", reader.count_lines())
print("Words:", reader.count_words())
print("Characters:", reader.count_characters())

reader.display_content()
```
