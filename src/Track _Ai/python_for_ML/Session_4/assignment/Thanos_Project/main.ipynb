```python
import os
import random
```

```python
def create_folder():
    os.mkdir("Thanos")
```

```python
create_folder()
```

```python
def create_files():
    num = int(input("Enter number of files: "))

    for i in range(1, num + 1):
        with open(f"Thanos/file{i}.txt", "w") as file:
            pass
```

```python
create_files()
```

```python
files = os.listdir("Thanos")

print(files)
```

```python
print(len(files))
```

```python
delete_files = random.sample(files, len(files)//2)
```

```python
print(delete_files)
```

```python
for file in delete_files:
    os.remove(f"Thanos/{file}")
```
