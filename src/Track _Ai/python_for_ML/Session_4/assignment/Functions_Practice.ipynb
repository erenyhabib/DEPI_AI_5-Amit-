Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime then they are known as twin primes

```python
def is_prime(number):

    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True


print("Twin Prime:")

for i in range(2, 1000):

    if is_prime(i) and is_prime(i + 2):
        print(f"{i} and {i + 2}")
```

4. Write a function that converts a decimal number to binary number

```python
def decimal_to_binary():

    num = int(input("Enter decimal number: "))
    binary = []
    while num > 0:
        remainder = num % 2
        binary.append(remainder)
        num = num // 2

    binary.reverse()

    print("Binary number:")

    for i in binary:
        print(i, end="")


decimal_to_binary()
```

5.A number is called perfect

```python
def is_perfect(number):

    sum_of_factors = 0
    for i in range(1, number):
        if number % i == 0:
            sum_of_factors += i

    if sum_of_factors == number:
        return True

    else:
        return False


start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))

print("Perfect Numbers:")

for i in range(start, end + 1):

    if is_perfect(i):
        print(i)
```
