```python
from abc import ABC ,abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass
    
class Bird(Animal):
    def move(self):
        print("move from bird")
        #pass
class Cat(Animal):
    def move(self):
        print("move from cat")
    
a=Bird()
a.move()
```

```python
class Employee:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    def set_salary(self,obj,new_sal):
        obj.salary=new_sal
        
```

```python
class Manager:
    def __init__(self,name,salar):
            self.name=name
            self._salary=salar
```

```python
e1=Employee("ereny",2000)
m1=Manager("Ahmed",5000)
e1.set_salary(m1,6000)
print(m1.salary)
```
