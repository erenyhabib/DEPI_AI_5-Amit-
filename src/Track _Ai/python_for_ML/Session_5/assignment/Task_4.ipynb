```python
class DecimalToBinary:
    """
    Converts a decimal number to binary number.
    """

    def __init__(self, number):
        self.number = number

    def decimal_to_binary(self):
        num = self.number
        binary = []

        while num > 0:
            remainder = num % 2
            binary.append(remainder)
            num = num // 2

        binary.reverse()

        print("Binary number:")

        for i in binary:
            print(i, end="")


```
