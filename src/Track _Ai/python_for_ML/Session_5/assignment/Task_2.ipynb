```python
class TwinPrimes:
    """
    Handles finding twin prime numbers.
    """

    def __init__(self, limit):
        self.limit = limit

    def is_prime(self, number):
        """
        Checks whether a number is prime.
        """

        if number < 2:
            return False

        for i in range(2, number):
            if number % i == 0:
                return False

        return True

    def find_twin_primes(self):
        """
        Finds and prints twin prime numbers within the given limit.
        """

        print("Twin Prime:")

        for i in range(2, self.limit):
            if self.is_prime(i) and self.is_prime(i + 2):
                print(f"{i} and {i + 2}")


        
```
