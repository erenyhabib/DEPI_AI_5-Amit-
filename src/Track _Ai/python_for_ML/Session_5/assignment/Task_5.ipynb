```python
class PerfectNumber:
    """
    Handles finding perfect numbers in a given range.
    """

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def is_perfect(self, number):
        """
        Checks whether a number is a perfect number.
        """

        sum_of_factors = 0

        for i in range(1, number):
            if number % i == 0:
                sum_of_factors += i

        if sum_of_factors == number:
            return True
        else:
            return False

    def find_perfect_numbers(self):
        """
        Finds and prints perfect numbers in the given range.
        """

        print("Perfect Numbers:")

        for number in range(self.start, self.end + 1):
            if self.is_perfect(number):
                print(number)


 
```

```python
start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))

task5 = PerfectNumber(start, end)

task5.find_perfect_numbers()   
```
