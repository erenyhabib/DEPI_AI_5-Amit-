```python
class PrimeFactors:
    """
        Handles finding prime  factor of number.
    """
    def __init__(self, number):
        self.number=number
        
    def find_prime_factors(self):
        """
        Finds and prints the prime factors of the number.
        """

        factors = []
        number = self.number

        for i in range(2, number + 1):
            while number % i == 0:
                factors.append(i)
                number = number // i

        print("Prime factors:", factors)
```

```python
number = int(input("Enter a number: "))

task3 = PrimeFactors(number)

task3.find_prime_factors()
```
