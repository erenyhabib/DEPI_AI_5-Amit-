```python
import os
import random


class Thanos:
    """
    Manages files inside a folder and randomly deletes half of them.
    """

    def __init__(self, folder, number_of_files):
        self.folder = folder
        self.number_of_files = number_of_files

    def create_files(self):
        """
        Creates the required number of files inside the folder.
        """
        os.makedirs(self.folder, exist_ok=True)

        for i in range(1, self.number_of_files + 1):
            with open(f"{self.folder}/file{i}.txt", "w") as file:
                pass

    def count_files(self):
        """
        Counts the files inside the folder.
        """
        return len(os.listdir(self.folder))

    def delete_random_half(self):
        """
        Deletes a random half of the files in the folder.
        """
        files = os.listdir(self.folder)

        random.shuffle(files)

        half = len(files) // 2

        for file in files[:half]:
            os.remove(os.path.join(self.folder, file))

    def run(self):
        """
        Runs the Thanos project.
        """
        self.create_files()

        print("Files before Snap:", self.count_files())

        self.delete_random_half()

        print("Files after Snap:", self.count_files())
```

```python
folder = "Thanos"
number = int(input("Enter number of files: "))

thanos = Thanos(folder, number)

thanos.run()
```
