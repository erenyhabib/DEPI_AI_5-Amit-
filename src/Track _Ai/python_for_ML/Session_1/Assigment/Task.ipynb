```python
email = "Amit_ml@gmail.edu"
```

```python
first_check=email.count("@")
print(f"the count of @ is {first_check}")
```

```python
second_check=email.count(".")
print(f"the count of . is {second_check}")
```

```python
if first_check==1 and second_check==1:
    print("email is valid")
else:
    print("email Invalid")
```

```python
index=email.find('.')
index
```

```python
Extract_Domain=email[find_index+1:index]
print(Extract_Domain)
```

```python
end_email=email.endswith(".edu")
end_email
```

```python
if end_email==True:
    print("email is education")
else:
    print("Commercial Domain")
```
